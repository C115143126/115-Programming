#  簡介

串列(list)是一個可以儲存多個元素(整數、浮點數或字串)的容器。  

串列的內容以逗點隔開左右有中括號[]


```python
score = [78,29,81,93]
print(score)
```

##  索引值
可以使用索引值(index)來存取串列中的元素。  

索引值從0開始

```python
print(score[0])
print(score[1])
```

##  串列內容的個數

len(a)取得 a 這個元素的個數。  

len 是 lenght(長度)的意思

```python
print(len(score))
print(score[3])#最後一個數
print(score[len(score)-1])
print(score[-1])
```

##  新增元素

使用`append()`新增元素到串列尾端

e.x. score.append(70)

```python
print(score)
score.append(70)
score.append(53)
print(score)
```

##  走訪串列的三種寫法



```python
print(score)
for i in range(len(score)):
    print(score[i])
```

```python
#對於score中的每個元素，一次取一個，放在e中
for e in score:
    print(e)
```

```python
for i, e in enumerate(score):
    print(i, e)
```

```python
#初始化串列，有91個0
fib = [0]*91
fib[0] = 1
fib[1] = 1
for i in range(2,91):
    fib[i] = fib[i-1] + fib[i-2]
print(fib[90])

```
