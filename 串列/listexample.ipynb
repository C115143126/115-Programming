#  簡介

串列(list)是一個可以儲存多個元素(整數、浮點數或字串)的容器。  

串列的內容以逗點隔開左右有中括號[]


```python
score = [78,29,81,93]
print(score)
```

##  索引值
可以使用索引值(index)來存取串列中的元素。  

索引值從0開始

```python
print(score[0])
print(score[1])
```

##  串列內容的個數

len(a)取得 a 這個元素的個數。  

len 是 lenght(長度)的意思

```python
print(len(score))
print(score[3])#最後一個數
print(score[len(score)-1])
print(score[-1])
```
