# Loop

Loop指令可以讓程式碼依我的條件不斷地進行，直到條件滿足為止。

```python
# 底下的程式碼會印出 0-9

for i in range(1,11,1):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i+1)
```

# 99乘法表

1\*1=1 1\*2=2 1\*3=3 ...  
2\*1=2 2\*2=4 2\*3=6 ...  
3\*1=3 3\*2=6 3\*3=9 ...  

```python
j=1
print(f"1*{j}={1*j}", end=",")
j=2
print(f"1*{j}={1*j}", end=",")
j=3
print(f"1*{j}={1*j}", end=",")
```

```python

for i in range(1,10,1):
    for j in range(1,10,1):

        print(f"{i}*{j}={i*j}", end=",")
    print()
```

#  練習

將上面最後一個逗號去掉  
用 if


```python

for i in range(1,10,1):
    for j in range(1,10,1):
        if(j==9):
        print(f"{i}*{j}={i*j}", end=",")
    print()


```
