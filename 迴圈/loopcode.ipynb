# Loop

Loop指令可以讓程式碼依我的條件不斷地進行，直到條件滿足為止。

```python
# 底下的程式碼會印出 0-9

for i in range(1,11,1):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i+1)
```
